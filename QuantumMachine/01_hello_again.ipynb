```python
with program() as prog:
    with infinite_loop_():
        play('trigger','trigger')
        # Pulse sequence
        play('pulse','qubit',duration=400*u.ns)
        wait(200*u.ns)
        play('pulse','qubit',duration=400*u.ns)
        wait(10*u.us)
```
